```python
import requests
```

```python
facts = []

for i in range(3):
    response = requests.get("https://catfact.ninja/fact")
    facts.append(response.json()["fact"])

print(facts)
```

```python
for index, fact in enumerate(facts,start=1):
    print(f"<h{index}> {fact}  </h{index}>")
```

```python
fact_tags = ""
for index, fact in enumerate(facts,start=1):
    fact_tags +=  f"<h{index}> {fact}  </h{index}>\n"
print(fact_tags)
```

```python
HTML_CONTENT = f"""<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
</head>
<body>
    {fact_tags}
</body>
</html>"""
```

```python
print(HTML_CONTENT)
```

```python
with open("cat.html", "w") as file_writer:
    file_writer.write(HTML_CONTENT)
```
