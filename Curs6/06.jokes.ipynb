```python
import requests
```

```python
BASE_URL = "https://v2.jokeapi.dev/joke/"
```

```python
jokes = {"Christmas":"red", "Programming":"green", "Dark":"black"}
```

```python
joke_dict = {}
for joke_type in jokes:
    response = requests.get(BASE_URL+joke_type, params={"format":"txt"})
    print(response.text)
    joke_dict[joke_type] = response.text
```

```python
joke_dict
```

```python
body_content = ""
for cat in joke_dict:
    body_content += f"<div id={cat}>  {joke_dict[cat]}  </div>"
print(body_content)
```

```python
style_content = ""
for cat in jokes:
    style_content += f"""#{cat} {{
        color: {jokes[cat]}
    }}
"""
    
print(style_content)
```

```python
with open("06.jokes.html", "r") as file_reader:
    html_content = file_reader.read()
```

```python
with open("06.jokes.html", "w") as file_writer:

    file_writer.write(html_content.replace("<style>", "<style>"+style_content).replace("<body>", "<body>"+body_content))
```
