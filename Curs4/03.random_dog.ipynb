```python
import requests
```

```python
response = requests.get("https://dog.ceo/api/breeds/image/random")
```

```python
next_url = response.json()['message']
```

```python
response = requests.get(next_url)
```

```python
dog_file_name = next_url.rsplit("/", 1)[-1]
dog_file_name
```

```python
with open( dog_file_name, "wb") as file_writer:
    file_writer.write(response.content)
```
