```python
import requests
import csv
```

```python
response = requests.get("https://randomuser.me/api/?format=csv&results=3").text
print(response)
```

```python
with open("users.csv", "w") as file_writer:
    file_writer.write(response)
```

```python
with open("users.csv", "r") as file_reader:
    for line in csv.DictReader(file_reader):
        print(type(line))
```
