```python
import requests
```

```python
headers = {
    "Accept" : "application/json"
}

response = requests.get("https://icanhazdadjoke.com", headers=headers)
response.json()
```

```python
response_dict = response.json()
joke_id = response_dict["id"]
```

```python
response2 = requests.get(f"https://icanhazdadjoke.com/j/{joke_id}.png", headers=headers)
response2
```

```python
with open(f"joke_{joke_id}.png", "wb") as file_writer:
    file_writer.write(response2.content)
```
