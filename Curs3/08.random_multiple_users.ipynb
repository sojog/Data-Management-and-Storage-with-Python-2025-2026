```python
import requests
```

```python
response = requests.get("https://randomuser.me/api/?results=3").json()
response
```

```python
len(response["results"])
```

```python
titles = {
    "Mr" : "Dl",
    'Monsieur' : "Dl",
    "Ms" : "Dna",
    "Miss" : "Dna",
    "Mrs" : "Dna",
}
```

```python
for result in response["results"]:
    user = result["name"]
    user_title = titles.get(user["title"], "_")
    print(f"{user["title"]} {user_title} {user["first"]} {user["last"]}")
```
