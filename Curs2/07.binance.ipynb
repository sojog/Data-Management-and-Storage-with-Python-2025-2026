```python
import requests
import random
```

```python
subdomains =  ["", "-gpc", "1", "2", "3", "4"]
sub = random.choice(subdomains)
```

```python
BASE_URL = f"https://api{sub}.binance.com"
BASE_URL
```

```python
PRICE_PATH = "/api/v3/avgPrice"
```

```python
URL = BASE_URL + PRICE_PATH # + "?symbol=ETHUSDT"
```

```python
params = {"symbol":"ETHUSDT"}
```

```python
response = requests.get(URL, params=params)
response.json()
```
