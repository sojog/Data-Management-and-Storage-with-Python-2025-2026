```python
import requests
```

```python
BASE_URL = "https://country.io"
```

```python
NAMES_PATH = "/names.json"
CAPITALS_PATH = "/capital.json"
```

```python
names_reponse =  requests.get(BASE_URL + NAMES_PATH)
capitals_response = requests.get(BASE_URL + CAPITALS_PATH)
```

```python
names_dict = names_reponse.json()
capital_dict = capitals_response.json()
```

```python
inverted_dict = {}
for short, long in names_dict.items():
    inverted_dict[long] = capital_dict[short]

inverted_dict
    
```

```python
while True:
    tara = input("La ce tara vrei sa afli capitala")
    print(f"Capitala {tara} este {inverted_dict[tara]}")
```
