```python
import requests
import json
from pprint import pprint
```

```python
BASE_URL = "https://api.chucknorris.io/jokes/"

CATEGORIES_URL =  BASE_URL + "categories"

SEARCH_PARTICULAR_CATEGORY = BASE_URL +  "search?query=" 
RANDOM_PARTICULAR_CATEGORY =  BASE_URL +"random?category="
```

```python
response = requests.get(CATEGORIES_URL)
JOKES_CATEGORIES = response.json()
JOKES_CATEGORIES
```

```python
JOKES_PER_CATEGORY_DICT = {

}
```

```python
for cat in JOKES_CATEGORIES:
    response = requests.get(RANDOM_PARTICULAR_CATEGORY+cat)
    JOKES_PER_CATEGORY_DICT[cat] = response.json()["value"]
    pprint(response.json())
```

```python
JOKES_PER_CATEGORY_DICT
```

```python
with open("jokes_cat.json", "w") as file_writer:
    json.dump(JOKES_PER_CATEGORY_DICT, file_writer)
```
