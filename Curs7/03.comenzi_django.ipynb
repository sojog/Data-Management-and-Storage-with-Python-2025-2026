 pip install Django

# 2. Creare proiect django (numele "proiectdjango" poate fi inlocuit cu altceva)

python -m django startproject proiectdjango

# 3. Pornire server        (trebuie sa fiu in folder in care este si manage.py)
cd proiectdjango 


python manage.py runserver
