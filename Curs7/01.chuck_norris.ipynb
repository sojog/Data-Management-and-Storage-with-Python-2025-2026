```python
import requests
import json
import pprint
```

```python
BASE_URL = "https://api.chucknorris.io/jokes/"

POLITICAL_URL =  BASE_URL + "random?category=political"

CHINA_URL = BASE_URL +  "search?query=china"
```

```python
response = requests.get(POLITICAL_URL)
response.json()
```

```python
political_joke = response.json()["value"]
political_joke
```

```python
response = requests.get(CHINA_URL)
result_jokes = response.json()['result']
result_jokes
```

```python
result_jokes[0]["value"]
```

```python
for joke_dict in result_jokes:
    if "political" in joke_dict["categories"]:
        print(joke_dict["value"])
        break
```
